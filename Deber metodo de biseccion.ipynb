-- Método de bisección

```python

#metodo de biseccion
def biseccion(f, a, b, TOL=1e-6, N0=100):
    FA = f(a)
    historia = []
    i = 1

    while i <= N0:
        p = (a + b) / 2
        FP = f(p)
        historia.append((a, b, p))

        if FP == 0 or (b - a) / 2 < TOL:
            return p, historia

        i += 1

        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

    raise RuntimeError(f"Fracaso después de {N0} iteraciones.")
```

1. CONDICIONES Y COTA DE ITERACIONES. La ecuación $x^3 - x - 1 = 0$ tiene una solución en $[1,2]$. Se busca una aproximación con precisión de $10^{-4}$.

```python


def f(x):
    return x**3 - x - 1


p, historia = biseccion(f, 1.0, 2.0, TOL=1e-4)


print(f"Solución encontrada: {p:.6f}")
print(f"Total de iteraciones: {len(historia)}\n")
print(f"{'i':>3} | {'a':>10} | {'b':>10} | {'p':>10} | {'f(p)':>12} | {'(b-a)/2':>10}")
print("-" * 65)

for i, (a_i, b_i, p_i) in enumerate(historia, 1):
    fp = f(p_i)
    p = (b_i - a_i) / 2
    print(f"{i:3d} | {a_i:10.6f} | {b_i:10.6f} | {p_i:10.6f} | {fp:12.4e} | {p:10.6f}")
```

¿A qué cero converge la bisección en cada intervalo? Escriba los ceros en este orden: $[-3.5, 2.5]$, $[-1.5, 0.6]$, $[0.4, 2.2]$, $[2, 3.7]$.

```python

def f(x):
    return (x + 3) * (x + 1)**2 * x * (x - 1)**3 * (x - 3)

intervalos = [
    (-3.5, 2.5),
    (-1.5, 0.6),
    (0.4, 2.2),
    (2.0, 3.7)
]

for a, b in intervalos:
    p, historia = biseccion(f, a, b, TOL=1e-6)
    print(f"Intervalo [{a}, {b}] -> Converge al cero: {round(p)}")
```

4.PROBLEMAS APLICADOS Un abrevadero de longitud $L$ tiene una sección transversal en forma de semicírculo de radio $r$. Cuando se llena de agua hasta una distancia $h$ desde la parte superior, el volumen de agua es

$$V = L\left[0.5\pi r^2 - r^2 \arcsin\left(\frac{h}{r}\right) - h\left(r^2 - h^2\right)^{1/2}\right].$$

Abrevadero de sección semicircular

Suponga que $L = 10$ cm, $r = 1$ cm y $V = 12.4\ \text{cm}^3$.

```python
import numpy as np

L = 10.0
r = 1.0
V = 12.4

def f_abrevadero(h):
    return L * (0.5 * np.pi * r**2 - r**2 * np.arcsin(h / r) - h * np.sqrt(r**2 - h**2)) - V

h_aprox, historia = biseccion(f_abrevadero, 0.0, 1.0, TOL=0.01)

profundidad = r - h_aprox

print(f"Iteraciones realizadas: {len(historia)}")
print(f"Distancia h encontrada: {h_aprox:.6f} cm (o {h_aprox:.2f} cm)")
print(f"Profundidad del agua:   {profundidad:.6f} cm (o {profundidad:.2f} cm)\n")


print(f"{'i':>2} | {'a':>8} | {'b':>8} | {'p (h)':>8} | {'(b-a)/2':>8}")
print("-" * 45)
for i, (a_i, b_i, p_i) in enumerate(historia, 1):
    print(f"{i:2d} | {a_i:8.4f} | {b_i:8.4f} | {p_i:8.4f} | {(b_i - a_i)/2:8.4f}")
```

Un objeto de masa $m$ cae desde una altura $s_0$, sujeto a la gravedad y a una resistencia viscosa. Su altura después de $t$ segundos es

$$s(t) = s_0 - \frac{mg}{k}\,t + \frac{m^2 g}{k^2}\left(1 - e^{-kt/m}\right),$$

con $g = 9.81\ \text{m/s}^2$ y $k$ el coeficiente de resistencia del aire, en $\text{N}\cdot\text{s/m}$. Suponga $s_0 = 300$ m, $m = 0.25$ kg y $k = 0.1\ \text{N}\cdot\text{s/m}$.

```python
def biseccion(f, a, b, TOL=1e-6, N0=100):
    """Devuelve (p, historia), con historia = [(a, b, p) por iteración]."""

    FA = f(a)
    historia = []  
    i = 1

    while i <= N0:
        p = (a + b) / 2
        FP = f(p)
        historia.append((a, b, p))

        if FP == 0 or (b - a) / 2 < TOL:
            return p, historia

        i += 1

        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

    raise RuntimeError(f"Fracaso después de {N0} iteraciones.")


g = 9.81
s0, m, k = 300.0, 0.25, 0.1


def s(t):
    return s0 - (m * g / k) * t + (m ** 2 * g / k ** 2) * (1 - np.exp(-k * t / m))


def graficar(t_star, historia, k_iteracion=None, archivo="caida.png"):
    """Dibuja s(t). Si k_iteracion no es None, resalta el intervalo [a, b]
    de esa iteración (1-based) y su punto medio p."""
    t = np.linspace(0, 20, 400)
    plt.figure(figsize=(9, 4.5))
    plt.plot(t, s(t), label="s(t)")
    plt.axhline(0, color="gray", linestyle="--", linewidth=1)
    if k_iteracion is not None:
        a, b, p = historia[k_iteracion - 1]
        plt.axvspan(a, b, color="green", alpha=0.25,
                    label=f"iter {k_iteracion}: [{a:.4g}, {b:.4g}]")
        plt.axvline(p, color="green", linestyle=":", linewidth=1,
                    label=f"p = {p:.4g}")
    plt.plot(t_star, 0, "ro", label=f"t* = {t_star:.6f}")
    plt.xlabel("t (s)")
    plt.ylabel("s(t) (m)")
    plt.title("Caída con resistencia del aire")
    plt.legend()
    plt.grid(True)
    plt.savefig(archivo, dpi=150)
    plt.show()


t_star, historia = biseccion(s, 0.0, 20.0, TOL=1e-4)
graficar(t_star, historia, k_iteracion=3)
```
